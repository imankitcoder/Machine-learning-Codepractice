```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error,root_mean_squared_error
```

```python
df = pd.read_csv("Ice_cream selling data.csv")
df
```

```python
df.corr()
```

```python
x = df[['Temperature (°C)']]
y = df['Ice Cream Sales (units)']
```

```python
res = train_test_split(x,y,train_size=0.8)
print(type(res),len(res))
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,train_size=0.8,random_state=42)
```

## Actual Data vs Model predicted data

```python
ytrain_predicted = model.predict(xtrain)
```

```python
plt.scatter(xtrain,ytrain)  #actual input vs actual output
plt.scatter(xtrain, ytrain_predicted)   #actual input vs predicted output
```

Conclusion:
Model is not able to captures the patterns.

Reason:
The data is non-linear and linear regression can capture only linear data.

```python
'''
Why we're comparing this?
1) plt.scatter(xtrain,ytrain) #actual input vs actual output.
2) plt.scatter(xtrain,ytrain_predicted) # actual input vs predicted output.

Answer:
From this comparison between actual pattern and predicted pattern, we can conclude how well the model captures the pattern during model training????

'''
```

```python
ytest_predicted = model.predict(xtest)
```

```python
plt.scatter(xtest,ytest)  #actual input vs actual output
plt.scatter(xtest, ytest_predicted)   #actual input vs predicted output
```

```python
''' 
From this plot we can predict that the predicted output is completely different from the actual output. This is because the data is non-linear. So, we cannot use the linear regression for this data.

Model fails to capture the pattern from the unseened data hence it will not give the proper future prediction.

Without training, it is not possible to access the intercept and slope.
model is the var here that stores the intercept and the slope data.
if we comment the model it will give attribute error.

'''
```

```python
train_data = pd.DataFrame({
    'Xtrain':xtrain['Temperature (°C)'],
    'Ytrain':ytrain,
    'Ytrain_pred': ytrain_predicted
})
train_data['Error'] = train_data['Ytrain']-train_data['Ytrain_pred']
```

### MSE ( mean squared error )

```python
from sklearn.metrics import mean_squared_error
```

```python
train_mse = mean_squared_error(ytrain,ytrain_predicted)
test_mse = mean_squared_error(ytest,ytest_predicted)

print(train_mse,test_mse)
```

## Model evaluation


```python
## r2 score
## Train score

from sklearn.metrics import r2_score

r2_train_score = r2_score(ytrain,ytrain_predicted)
r2_test_score = r2_score(ytest,ytest_predicted)
print(r2_train_score)
print(r2_test_score)
```

Model is not able to capture train or actual pattern 

Why?

Because the data is non linear

model is not ready for the future prediction

or

model fails to captures unseen data pattern

```python
## MAE

train_score_mae = mean_absolute_error(ytrain,ytrain_predicted)
test_score_mae = mean_absolute_error(ytest,ytest_predicted)
print(train_score_mae)
print(test_score_mae)
```

```python
#RMSE

train_score_rmse = root_mean_squared_error(ytrain,ytrain_predicted)
test_score_rmse = root_mean_squared_error(ytest, ytest_predicted)
print(train_score_rmse)
print(test_score_rmse)
```
